```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # Approximate radius of Earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# 1. Define launch site coordinates (CCAFS SLC-40) and coastline target coordinates
launch_site_lat = 28.563197
launch_site_lon = -80.576820
coastline_lat = 28.56367
coastline_lon = -80.57163

# 2. Calculate distance
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)

# 3. Create a Folium Marker with distance text label
distance_marker = folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html=f'<div style="font-size: 12px; color:#d35400;"><b>{distance_coastline:.2f} KM</b></div>'
    )
)
site_map.add_child(distance_marker)

# 4. Draw PolyLine between launch site and coastline point
lines = folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]],
    weight=1,
    color='blue'
)
site_map.add_child(lines)

site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
# 1. Define launch site coordinates (CCAFS SLC-40)
launch_site_lat, launch_site_lon = 28.563197, -80.576820
launch_coordinate = [launch_site_lat, launch_site_lon]

# 2. Define coordinates for closest infrastructure targets
proximities = {
    'Railway': [28.57206, -80.58528],
    'Highway': [28.56321, -80.57079],
    'City': [28.61200, -80.80700]
}

# 3. Loop through targets to calculate distance, add markers, and draw lines
for name, target_coord in proximities.items():
    # Calculate distance using Haversine formula
    distance = calculate_distance(launch_site_lat, launch_site_lon, target_coord[0], target_coord[1])
    
    # Create distance marker
    distance_marker = folium.Marker(
        target_coord,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html='<div style="font-size: 12px; color:#d35400;"><b>%s</b></div>' % "{}: {:10.2f} KM".format(name, distance)
        )
    )
    site_map.add_child(distance_marker)
    
    # Create polyline connecting launch site to target point
    lines = folium.PolyLine(
        locations=[launch_coordinate, target_coord],
        weight=1,
        color='blue'
    )
    site_map.add_child(lines)

site_map
```
